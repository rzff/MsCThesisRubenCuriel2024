```python
import pandas as pd
from pathlib import Path
from typing import Tuple, Union
```

```python
def split_model_results(
    data: Union[str, Path, pd.DataFrame]
) -> Tuple[pd.DataFrame, pd.DataFrame]:
        # Load data
    if not isinstance(data, pd.DataFrame):
        df = pd.read_csv(data)
    else:
        df = data.copy()
    
    # Normalize model type
    df['model_type'] = df['model_type'].str.upper()
    
    # Identify grouping keys
    keys = [c for c in df.columns if c.lower() in ('fold', 'horizon', 'model_type')]
    
    # Metrics are everything else
    metric_cols = [c for c in df.columns if c not in keys]
    
    # Split
    df_lstm  = df[df['model_type'] == 'LSTM'].copy()
    df_qlstm = df[df['model_type'] == 'QLSTM'].copy()
    
    # Rename metrics
    df_lstm  = df_lstm.rename(columns={c: f"LSTM_{c}"   for c in metric_cols})
    df_qlstm = df_qlstm.rename(columns={c: f"QLSTM_{c}"  for c in metric_cols})
    
    return df_lstm.reset_index(drop=True), df_qlstm.reset_index(drop=True)
    
```

```python
# ------------------------------------------------------------------
# helper: mean + 95 % CI
# ------------------------------------------------------------------
def mean_ci(series):
    mean = series.mean()
    sem  = series.std(ddof=1) / np.sqrt(len(series))
    return mean, 1.96 * sem

# ------------------------------------------------------------------
# model-column mapping (unchanged)
# ------------------------------------------------------------------
mapping = {
    'Prophet LSTM':           'Prophet LSTM',
    'Prophet QLSTM':          'Prophet QLSTM',
    'Stacked Prophet LSTM':   'Stacked Prophet LSTM',
    'Stacked Prophet QLSTM':  'Stacked Prophet QLSTM',
}
metrics = ['RMSE', 'MAPE', 'PCC']

# ------------------------------------------------------------------
# aggregate fold-level stats (unchanged)
# ------------------------------------------------------------------
agg_rows = []
for model, prefix in mapping.items():
    row = {'Model': model}
    for metric in metrics:
        col = f'{prefix} {metric}'
        mean, ci = mean_ci(df[col])
        row[f'{metric}_Mean'] = mean
        row[f'{metric}_CI']   = ci
    agg_rows.append(row)
agg_df = pd.DataFrame(agg_rows)

# baseline MAPE
baseline_mean, baseline_ci = mean_ci(df['Prophet Baseline MAPE'])

# ------------------------------------------------------------------
# plotting helper — horizontal bars
# ------------------------------------------------------------------
import matplotlib.pyplot as plt
fill_color = 'orange'
edge_color = 'black'
line_width = 1.2
dpi = 300

def barh_plot(means, cis, labels, fname, xlabel):
    plt.figure(figsize=(6, 3.5))
    y_pos = np.arange(len(labels))
    plt.barh(
        y_pos, means, xerr=cis,
        color=fill_color, edgecolor=edge_color,
        linewidth=line_width, capsize=4, align='center'
    )
    plt.yticks(y_pos, labels)
    plt.xlabel(xlabel)
    # Remove the next line if you want "best" at the bottom
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.savefig(fname, dpi=dpi, bbox_inches='tight')
    plt.close()

# ------------------------------------------------------------------
# generate charts
# ------------------------------------------------------------------
# 1. RMSE
barh_plot(
    agg_df['RMSE_Mean'], agg_df['RMSE_CI'],
    agg_df['Model'], 'rmse_chart.png', 'RMSE (MWh)'
)

# 2. MAPE (includes baseline)
mape_labels = list(agg_df['Model']) + ['Prophet Baseline']
mape_means  = list(agg_df['MAPE_Mean']) + [baseline_mean]
mape_cis    = list(agg_df['MAPE_CI'])   + [baseline_ci]
barh_plot(
    mape_means, mape_cis,
    mape_labels, 'mape_chart.png', 'MAPE (%)'
)

# 3. PCC
barh_plot(
    agg_df['PCC_Mean'], agg_df['PCC_CI'],
    agg_df['Model'], 'pcc_chart.png', 'PCC'
)
```

```python

from scipy.stats import ttest_rel
metrics_info = {
    'rmse':        ['cls_rmse', 'qlstm_rmse', 'LSTM_rmse', 'QLSTM_rmse'],
    'mape (%)':    ['cls_mape', 'qlstm_mape', 'LSTM_mape (%)', 'QLSTM_mape (%)'],
    'pearson_corr':['cls_pcc',  'qlstm_pcc',  'LSTM_pearson_corr',  'QLSTM_pearson_corr']
}

results = []
for metric, cols in metrics_info.items():
    model_names = ['CLS', 'QLSTM', 'Stacked_LSTM', 'Stacked_QLSTM']
    for i in range(len(cols)):
        for j in range(i+1, len(cols)):
            col1, col2 = cols[i], cols[j]
            paired = df[[col1, col2]].dropna()
            if len(paired) < 2:
                continue
            # Perform paired t-test
            t_stat, p_val = ttest_rel(paired[col1], paired[col2])
            
            # Handle direction of mean_diff based on metric type
            if metric in ('rmse', 'mape (%)'):
                # Lower is better → diff = model1 - model2
                mean_diff = (paired[col1] - paired[col2]).mean()
                if p_val < 0.05:
                    winner = model_names[i] if mean_diff < 0 else model_names[j]
                else:
                    winner = 'None'
            else:
                # Higher is better (PCC) → diff = model2 - model1
                mean_diff = (paired[col2] - paired[col1]).mean()
                if p_val < 0.05:
                    winner = model_names[j] if mean_diff > 0 else model_names[i]
                else:
                    winner = 'None'
                        
            results.append({
                'metric': metric,
                'model_1': model_names[i],
                'model_2': model_names[j],
                't_statistic': t_stat,
                'p_value': p_val,
                'mean_diff_(1−2)': mean_diff,
                'winner': winner
            })
res_df = pd.DataFrame(results)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import shapiro, sem, t
import matplotlib.pyplot as plt

# Define your metric groups
metric_groups = {
    'rmse':        ['cls_rmse', 'qlstm_rmse', 'LSTM_rmse', 'QLSTM_rmse'],
    'mape (%)':    ['cls_mape', 'qlstm_mape', 'LSTM_mape (%)', 'QLSTM_mape (%)'],
    'pearson_corr':['cls_pcc',  'qlstm_pcc',  'LSTM_pearson_corr',  'QLSTM_pearson_corr']
}

alpha = 0.05
B = 20000  # number of bootstrap samples
results = []

for group, metrics in metric_groups.items():
    for metric in metrics:
        data = df[metric].dropna().values
        n = len(data)
        
        # Shapiro–Wilk normality test
        W, p_val = shapiro(data)
        normal = p_val > alpha
        
        # Compute mean and standard deviation
        mean_val = data.mean()
        std_val  = data.std(ddof=1)  # sample standard deviation
        
        # Compute CI
        if normal:
            # Parametric t-CI
            se_val = sem(data, ddof=1)
            tcrit = t.ppf(1 - alpha/2, df=n-1)
            lo, hi = mean_val - tcrit * se_val, mean_val + tcrit * se_val
            ci_method = 't-CI'
        else:
            # Bootstrap CI
            boot_means = np.array([
                np.mean(np.random.choice(data, size=n, replace=True))
                for _ in range(B)
            ])
            lo, hi = np.percentile(boot_means, [2.5, 97.5])
            ci_method = 'bootstrap'
        
        # Record result
        results.append({
            'group':      group,
            'metric':     metric,
            'mean':       mean_val,
            'std':        std_val,
            'shapiro_W':  W,
            'shapiro_p':  p_val,
            'normal?':    normal,
            'CI method':  ci_method,
            'CI lower':   lo,
            'CI upper':   hi
        })

# Create and display summary DataFrame
summary_df = pd.DataFrame(results)
```

```python
import os
import re
import glob

import pandas as pd
import matplotlib.pyplot as plt

def parse_true_loss_file(path):
    """
    Parse one .err file named "<jobid>_<fold>.err",
    extracting per-batch true losses and marking variant
    as 'classical' until an epoch reset, then 'quantum'.
    """
    fn = os.path.basename(path)
    base, _ = os.path.splitext(fn)
    # split only on the LAST underscore:
    jobid, fold_str = base.rsplit("_", 1)
    fold = int(fold_str)

    records = []
    text = open(path, 'rb').read().replace(b'\r', b'\n').decode('utf-8', 'ignore')
    lines = text.split('\n')

    prev_ep = None
    variant = 'classical'

    for L in lines:
        # detect epoch resets
        m_ep = re.search(r"Epoch\s+(\d+)", L)
        if m_ep:
            ep = int(m_ep.group(1))
            if prev_ep is not None and ep < prev_ep:
                variant = 'quantum'
            prev_ep = ep

        # capture true loss lines
        m_loss = re.search(r"([0-9]+)\/([0-9]+).*loss=([0-9]+\.[0-9]+)", L)
        if m_loss and prev_ep is not None:
            batch = int(m_loss.group(1))
            total = int(m_loss.group(2))
            loss = float(m_loss.group(3))
            records.append({
                'jobid': jobid,
                'fold': fold,
                'variant': variant,
                'epoch': prev_ep,
                'batch': batch,
                'total_batches': total,
                'loss': loss,
                'filename': fn
            })
    return records


def extract_and_save(err_dir, out_csv):
    """Walk err_dir for *.err, parse losses, save to CSV."""
    all_records = []
    for path in sorted(glob.glob(os.path.join(err_dir, '*.err'))):
        all_records.extend(parse_true_loss_file(path))

    if not all_records:
        raise RuntimeError("No true loss records found in directory.")

    df = pd.DataFrame(all_records)
    df.to_csv(out_csv, index=False)
    print(f"Saved consolidated loss CSV: {out_csv} ({len(df)} rows)")
    return out_csv


def plot_from_csv(csv_path):
    """Read CSV and plot end‐of‐epoch loss for each fold & variant."""
    df = pd.read_csv(csv_path)
    for col in ['fold', 'variant', 'epoch', 'batch', 'loss']:
        if col not in df.columns:
            raise RuntimeError(f"Missing '{col}' in CSV.")

    df[['epoch','batch','loss']] = df[['epoch','batch','loss']].apply(pd.to_numeric, errors='coerce')

    out_dir = os.path.dirname(csv_path) or '.'
    for fold in sorted(df['fold'].unique()):
        sub = df[df['fold'] == fold]
        plt.figure()
        for marker, style, variant in [('o-', 'Classical', 'classical'),
                                       ('x--', 'Quantum', 'quantum')]:
            vdf = sub[sub['variant'] == variant]
            if vdf.empty:
                continue
            idx = vdf.groupby('epoch')['batch'].idxmax()
            endpts = vdf.loc[idx].sort_values('epoch')
            plt.plot(endpts['epoch'], endpts['loss'], marker, label=style)

        plt.title(f'Fold {fold} End-of-Epoch Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.grid(True)
        plt.tight_layout()

        out_png = os.path.join(out_dir, f'loss_progression_fold_{fold}.png')
        plt.savefig(out_png)
        plt.close()
        print(f"Saved plot: {out_png}")


# ─── USER SETTINGS ──────────────────────────────────────────────────────────────
# Directory containing your .err files:
err_dir = "4230_StackedBoth/4230Loss"

# If you want to extract & plot in one go:
out_csv = "output_losses.csv"
csv_path = extract_and_save(err_dir, out_csv)
plot_from_csv(csv_path)

# OR, if you already have a CSV:
# csv_path = "path/to/existing_losses.csv"
# plot_from_csv(csv_path)
```

```python
import json, pandas as pd, matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import TwoSlopeNorm

# 1) ------------------------------------------------------------------
df = pd.read_json("feature_selection_results.json")

# 2) pivot to N_Climate × N_Econ --------------------------------------
pivot = (df.pivot_table(index="N_Climate", columns="N_Econ", values="RMSE")
            .sort_index(ascending=False))     # nicer orientation

# 3) blue-to-red palette ----------------------------------------------
# centre colour-map on the overall mean so “good blue / bad red” is balanced
norm = TwoSlopeNorm(vmin=pivot.min().min(),
                    vcenter=pivot.mean().mean(),
                    vmax=pivot.max().max())

plt.figure(figsize=(5.6, 4.4))
ax = sns.heatmap(pivot,
                 cmap="coolwarm_r",       # blue = low, red = high
                 norm=norm,
                 linewidths=.75, linecolor='white',
                 annot=True, fmt='.0f',   # show integer RMSE
                 cbar_kws=dict(label="RMSE (MWh)"))

ax.set_xlabel(r'$n_{\mathrm{econ}}$ (economic features)')
ax.set_ylabel(r'$n_{\mathrm{clim}}$ (climate features)')
ax.invert_yaxis()

# mark global optimum --------------------------------------------------
best = df.loc[df['RMSE'].idxmin()]
ax.plot(best['N_Econ'] + .5,
        best['N_Climate'] + .5,
        marker='*', color='black', markersize=18, markeredgecolor='yellow')

plt.tight_layout()
out_png = "media/plots/feature_heatmap_rmse_coolwarm.png"
plt.savefig(out_png, dpi=300)
plt.close()
print(f"✓ saved heat-map → {out_png}")
```
